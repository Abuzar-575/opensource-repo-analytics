```python
from pyspark.sql.types import *

user_struct = StructType([
    StructField("login", StringType(), True),
    StructField("id", LongType(), True),
    StructField("type", StringType(), True),
])

label_struct = StructType([
    StructField("id", LongType(), True),
    StructField("name", StringType(), True),
])

commit_person = StructType([
    StructField("name", StringType(), True),
    StructField("email", StringType(), True),
    StructField("date", StringType(), True),
])

commits_schema = StructType([
    StructField("repo_name", StringType(), True),
    StructField("entity_type", StringType(), True),
    StructField("sha", StringType(), True),
    StructField("commit", StructType([
        StructField("author", commit_person, True),
        StructField("committer", commit_person, True),
        StructField("message", StringType(), True),
        StructField("comment_count", LongType(), True),
    ]), True),
    StructField("author", user_struct, True),
    StructField("committer", user_struct, True),
])

issues_schema = StructType([
    StructField("repo_name", StringType(), True),
    StructField("entity_type", StringType(), True),
    StructField("id", LongType(), True),
    StructField("number", LongType(), True),
    StructField("title", StringType(), True),
    StructField("state", StringType(), True),
    StructField("user", user_struct, True),
    StructField("labels", ArrayType(label_struct), True),
    StructField("comments", LongType(), True),
    StructField("created_at", StringType(), True),
    StructField("updated_at", StringType(), True),
    StructField("closed_at", StringType(), True),
    StructField("pull_request", StructType([
        StructField("url", StringType(), True),
    ]), True),
])

pulls_schema = StructType([
    StructField("repo_name", StringType(), True),
    StructField("entity_type", StringType(), True),
    StructField("id", LongType(), True),
    StructField("number", LongType(), True),
    StructField("title", StringType(), True),
    StructField("state", StringType(), True),
    StructField("user", user_struct, True),
    StructField("created_at", StringType(), True),
    StructField("updated_at", StringType(), True),
    StructField("closed_at", StringType(), True),
    StructField("merged_at", StringType(), True),
    StructField("merged", BooleanType(), True),
    StructField("comments", LongType(), True),
    StructField("review_comments", LongType(), True),
    StructField("commits", LongType(), True),
    StructField("additions", LongType(), True),
    StructField("deletions", LongType(), True),
    StructField("changed_files", LongType(), True),
])

SCHEMAS = {"commits": commits_schema, "issues": issues_schema, "pulls": pulls_schema}

KEYS = {"commits": ["repo_name", "sha"],
        "issues":  ["repo_name", "id"],
        "pulls":   ["repo_name", "id"]}
```
