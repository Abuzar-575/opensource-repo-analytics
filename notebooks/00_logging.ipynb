```python
spark.sql("""
CREATE TABLE IF NOT EXISTS workspace.ops.pipeline_execution_logs (
    log_id           STRING,
    layer            STRING,
    entity           STRING,
    parameter        STRING,
    load_type        STRING,
    start_time       TIMESTAMP,
    end_time         TIMESTAMP,
    status           STRING,
    rows_inserted    LONG,
    rows_updated     LONG,
    error_message    STRING,
    load_timestamp   TIMESTAMP
) USING DELTA
""")
```

```python
import uuid
from datetime import datetime
from pyspark.sql import Row

def write_log(layer, entity, parameter, load_type, start_time,
              status, rows_inserted=0, rows_updated=0, error_message=None):
    row = Row(
        log_id=str(uuid.uuid4()), layer=layer, entity=entity,
        parameter=parameter, load_type=load_type,
        start_time=start_time, end_time=datetime.now(),
        status=status, rows_inserted=rows_inserted,
        rows_updated=rows_updated, error_message=error_message,
        load_timestamp=datetime.now()
    )
    (spark.createDataFrame([row], schema=spark.table("workspace.ops.pipeline_execution_logs").schema)
          .write.mode("append").saveAsTable("workspace.ops.pipeline_execution_logs"))
```
