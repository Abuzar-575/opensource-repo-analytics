```python
catalog = "workspace"

for schema in ["landing", "bronze", "silver", "ops"]:
    spark.sql(f"CREATE SCHEMA IF NOT EXISTS {catalog}.{schema}")

spark.sql(f"CREATE VOLUME IF NOT EXISTS {catalog}.landing.raw")

display(spark.sql(f"SHOW SCHEMAS IN {catalog}"))
```
