```python
   import subprocess
   SCRIPT = "/Volumes/workspace/landing/raw/extract_raw.py"

   def run_extract(*args):
       p = subprocess.Popen(["python", "-u", SCRIPT, *args],
                            stdout=subprocess.PIPE, stderr=subprocess.STDOUT, text=True)
       for line in p.stdout:
           print(line, end="")
       p.wait()
       print("exit code:", p.returncode)
```

```python
run_extract("--mode", "full",
            "--start", "2025-10-01T00:00:00Z", "--until", "2026-10-01T00:00:00Z",
            "--out", "/Volumes/workspace/landing/raw")
```
